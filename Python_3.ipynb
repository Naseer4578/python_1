```python
#print prime numbers between given two numbers

a=int(input())
b=int(input())
for i in range(a,b+1):
  for j in range(2,i):
   if(i%j==0):
    break
  else:
   print(i)
```

```python
#print the sum 0f the digits of a given numbers
n=int(input())
sum=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
print(sum)
```
