```python
#PROGRAM THAT COUNT THE NO OF VOWELS IN A STRING

a=str(input())
b=a.lower()
vowels="aeiou"
count_v=0
count_c=0
for ch in b:
  if ch.isalpha():
    if ch in vowels:
      count_v+=1
    else:
      count_c+=1
print("no of vowels",count_v)
print("no of consonants",count_c)
```

```python
#PROGRAM THAT CHECK SUB STRING IS PRSENT IN THE GIVEN STRING,IF PRESENT

s1 = str(input("Enter the main string: "))
s2 = str(input("Enter the substring to check: "))

if s2 in s1:
  print(f"'{s2 }' is present in '{s1}'")
else:
  print(f"'{s2}' is NOT present in '{s1}'")
```
